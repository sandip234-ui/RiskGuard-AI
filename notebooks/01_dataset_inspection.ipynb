```python
from pathlib import Path

TRAIN_PATH = Path("../data/raw/fraudTrain.csv")
TEST_PATH = Path("../data/raw/fraudTest.csv")

print(f"Train size: {TRAIN_PATH.stat().st_size / (1024**2):.2f} MB")
print(f"Test size:  {TEST_PATH.stat().st_size / (1024**2):.2f} MB")
```

```python
def fraud_distribution(path):
    counts = {0: 0, 1: 0}
    
    for chunk in pd.read_csv(path, usecols=["is_fraud"], chunksize=100_000):
        values = chunk["is_fraud"].value_counts()
        
        counts[0] += values.get(0, 0)
        counts[1] += values.get(1, 0)
    
    total = counts[0] + counts[1]
    
    return pd.Series({
        "legitimate": counts[0],
        "fraud": counts[1],
        "total": total,
        "fraud_percentage": counts[1] / total * 100
    })
```

```python
train_fraud = fraud_distribution(TRAIN_PATH)
test_fraud = fraud_distribution(TEST_PATH)

print("TRAIN")
display(train_fraud)

print("\nTEST")
display(test_fraud)
```

```python
# Load only the date column
train_dates = pd.read_csv(
    TRAIN_PATH,
    usecols=["trans_date_trans_time"]
)

test_dates = pd.read_csv(
    TEST_PATH,
    usecols=["trans_date_trans_time"]
)

train_dates["trans_date_trans_time"] = pd.to_datetime(
    train_dates["trans_date_trans_time"]
)

test_dates["trans_date_trans_time"] = pd.to_datetime(
    test_dates["trans_date_trans_time"]
)

print("TRAIN")
print("Start:", train_dates["trans_date_trans_time"].min())
print("End:  ", train_dates["trans_date_trans_time"].max())

print("\nTEST")
print("Start:", test_dates["trans_date_trans_time"].min())
print("End:  ", test_dates["trans_date_trans_time"].max())
```

```python
print("Unique cards in train:", 
      pd.read_csv(TRAIN_PATH, usecols=["cc_num"])["cc_num"].nunique())

print("Unique cards in test:", 
      pd.read_csv(TEST_PATH, usecols=["cc_num"])["cc_num"].nunique())
```

```python
train_cards = set(
    pd.read_csv(
        TRAIN_PATH,
        usecols=["cc_num"]
    )["cc_num"]
)

test_cards = set(
    pd.read_csv(
        TEST_PATH,
        usecols=["cc_num"]
    )["cc_num"]
)

overlap = train_cards & test_cards

print("Train cards:", len(train_cards))
print("Test cards:", len(test_cards))
print("Cards appearing in both:", len(overlap))
print(
    "Test cards seen during training:",
    len(overlap) / len(test_cards) * 100,
    "%"
)
```
